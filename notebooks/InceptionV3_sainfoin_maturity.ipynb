# InceptionV3 – maturity-stage classification of sainfoin from UAV RGB images

Transfer-learning model (InceptionV3, ImageNet weights) trained on the labelled image dataset (early / mature / late; 2,188 training, 269 validation and 269 test images) published at https://doi.org/10.5281/zenodo.24484581.

**How to run (Google Colab, GPU runtime):** put the dataset in Google Drive, set `DATA_DIR` in the configuration cell, and run all cells.

**Note on the saved outputs.** The outputs stored in this notebook are those of the original run used in the article. Comments, printed messages and plot labels were translated from Turkish into English afterwards, so some printed messages in the stored outputs are still in Turkish and the stored figures show Turkish class names (erken = early, gec = late, olgun = mature). No settings or code logic were changed. Results may differ slightly between runs because no global random seed was set.

```python
# Paths: set DATA_DIR to the folder that contains train/, validation/ and test/
# (image dataset: https://doi.org/10.5281/zenodo.24484581). Class folders: erken = early, gec = late, olgun = mature.
import os
DATA_DIR  = '/content/drive/MyDrive/dataset_sainfoin/ambarMa'
MODEL_DIR = '/content/drive/MyDrive/dataset_sainfoin/models'
os.makedirs(MODEL_DIR, exist_ok=True)
train_path      = os.path.join(DATA_DIR, 'train')
validation_path = os.path.join(DATA_DIR, 'validation')
test_path       = os.path.join(DATA_DIR, 'test')
CLASS_LABELS = ['early', 'late', 'mature']   # same (alphabetical) order as the folders erken, gec, olgun
```

```python
import tensorflow as tf
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications.inception_v3 import preprocess_input
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
import matplotlib.pyplot as plt
import datetime

# ---------------------------
# 1. Load datasets
# ---------------------------
img_height, img_width = 299, 299
batch_size = 32
numberOfClass = 3
AUTOTUNE = tf.data.AUTOTUNE

def load_ds(path, shuffle=True, seed=None):
    ds = tf.keras.preprocessing.image_dataset_from_directory(
        path,
        image_size=(img_height, img_width),
        batch_size=batch_size,
        shuffle=shuffle,
        seed=seed
    )
    return ds.prefetch(buffer_size=AUTOTUNE)

train_ds = load_ds(train_path, seed=125)
validation_ds = load_ds(validation_path, shuffle=False)
test_ds = load_ds(test_path, shuffle=False)

# ---------------------------
# 2. Data augmentation and preprocessing
# ---------------------------
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal_and_vertical"),
    tf.keras.layers.RandomRotation(0.3),
    tf.keras.layers.RandomZoom(0.3),
    tf.keras.layers.RandomContrast(0.3),
])

# Augmentation (training set only) and preprocessing
train_ds = train_ds.map(lambda x, y: (data_augmentation(x, training=True), y))
train_ds = train_ds.map(lambda x, y: (preprocess_input(x), y))
validation_ds = validation_ds.map(lambda x, y: (preprocess_input(x), y))
test_ds = test_ds.map(lambda x, y: (preprocess_input(x), y))

# ---------------------------
# 3. Model (transfer learning)
# ---------------------------
base_model = InceptionV3(
    include_top=False,
    input_shape=(img_height, img_width, 3),
    weights='imagenet'
)
base_model.trainable = False

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(numberOfClass, activation='softmax')
])

# ---------------------------
# 4. Callbacks
# ---------------------------
callbacks = [
    # stop training when val_loss stops improving
    EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True, verbose=1),

    # save the best model
    ModelCheckpoint(os.path.join(MODEL_DIR, "inception_best_v2.h5"), monitor='val_loss', save_best_only=True, verbose=1),

    # reduce the learning rate when val_loss plateaus
    ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-7, verbose=1)
]

# ---------------------------
# 5. Training, stage 1 (frozen backbone)
# ---------------------------
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Stage 1: training the classification head")
history = model.fit(
    train_ds,
    validation_data=validation_ds,
    epochs=20,  # upper limit; EarlyStopping restores the best epoch
    callbacks=callbacks,
    # no class weights were used for this model
)

# ---------------------------
# 6. Training, stage 2 (fine-tuning)
# ---------------------------
print("Stage 2: fine-tuning the last 50 layers")
base_model.trainable = True
for layer in base_model.layers[:-50]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_fine = model.fit(
    train_ds,
    validation_data=validation_ds,
    epochs=10,
    callbacks=callbacks,
    # no class weights were used for this model
)

# ---------------------------
# 7. Training curves
# ---------------------------
def plot_final_history(h1, h2):
    acc = h1.history['accuracy'] + h2.history['accuracy']
    val_acc = h1.history['val_accuracy'] + h2.history['val_accuracy']
    loss = h1.history['loss'] + h2.history['loss']
    val_loss = h1.history['val_loss'] + h2.history['val_loss']

    plt.figure(figsize=(14, 5))

    # Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(acc, label='Training accuracy')
    plt.plot(val_acc, label='Validation accuracy')
    plt.axvline(len(h1.history['accuracy'])-1, color='r', linestyle='--', label='Fine-tuning')
    plt.title('Accuracy')
    plt.legend()

    # Loss
    plt.subplot(1, 2, 2)
    plt.plot(loss, label='Training loss')
    plt.plot(val_loss, label='Validation loss')
    plt.axvline(len(h1.history['accuracy'])-1, color='r', linestyle='--', label='Fine-tuning')
    plt.title('Loss')
    plt.legend()
    plt.show()

plot_final_history(history, history_fine)

# ---------------------------
# 8. Evaluation on the test set
# ---------------------------
test_loss, test_acc = model.evaluate(test_ds)
print(f"Test accuracy: {test_acc:.4f}")
```

```python
# Training curves (loss and accuracy)
print(history.history.keys())
# dict_keys(['accuracy', 'loss', 'val_accuracy', 'val_loss'])

# Two panels (1 row, 2 columns)
fig, ax = plt.subplots(1, 2, figsize=(12, 6))

# Left: loss
ax[0].plot(history.history["loss"], label="training loss")
ax[0].plot(history.history["val_loss"], label="validation loss")
ax[0].set_title("Loss")
ax[0].legend()

# Right: accuracy
ax[1].plot(history.history["accuracy"], label="training accuracy")
ax[1].plot(history.history["val_accuracy"], label="validation accuracy")
ax[1].set_title("Accuracy")
ax[1].legend()

plt.tight_layout()
# save before plt.show()
plt.savefig("InceptionV3_loss_acc.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns

# 1. Get predictions on the test set
test_labels = []
test_preds = []

for images, labels in test_ds:
    preds = model.predict(images)
    test_labels.extend(labels.numpy())
    test_preds.extend(np.argmax(preds, axis=1))

# 2. Generate confusion matrix
cm = confusion_matrix(test_labels, test_preds)

# 3. Plot confusion matrix using seaborn
plt.figure(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap=sns.cubehelix_palette(as_cmap=True),
    xticklabels=CLASS_LABELS,
    yticklabels=CLASS_LABELS,
    annot_kws={"size": 16}
)

plt.xlabel('Predicted', size=16)
plt.ylabel('Actual', size=16)
plt.xticks(size=16)
plt.yticks(size=16)
plt.tight_layout()
# save before plt.show()
plt.savefig("InceptionV3_cm.png", dpi=300, bbox_inches="tight")
plt.show()

```

```python
from sklearn.metrics import classification_report, accuracy_score
import numpy as np

test_labels = np.array(test_labels)  # True labels
test_preds = np.array(test_preds)  # Predicted labels

# Calculate accuracy
accuracy = accuracy_score(test_labels, test_preds)
print("Accuracy:", accuracy)

# Generate classification report (Precision, Recall, F1-score)
class_names = CLASS_LABELS
report = classification_report(test_labels, test_preds, target_names=class_names)
print(report)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np

# Classification report
report = classification_report(
    test_labels,
    test_preds,
    target_names=class_names,
    output_dict=True
)

# Confusion matrix
cm = confusion_matrix(test_labels, test_preds)

# specificity per class
specificity = {}
for i, cls in enumerate(class_names):
    TN = cm.sum() - (cm[i, :].sum() + cm[:, i].sum() - cm[i, i])
    FP = cm[:, i].sum() - cm[i, i]
    specificity[cls] = TN / (TN + FP)

# classification report as text
report_str = classification_report(
    test_labels,
    test_preds,
    target_names=class_names
)

# append specificity
spec_str = "\nSpecificity:\n"
for cls in class_names:
    spec_str += f"{cls:10s}: {specificity[cls]:.2f}\n"

final_text = report_str + spec_str

# Plot
plt.figure(figsize=(9,7))
plt.text(0.01, 0.05, final_text, fontsize=12, fontfamily='monospace')
plt.axis('off')
plt.title('Classification Report with Specificity', fontsize=14)
plt.savefig("InceptionV3_acc.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Predict the maturity stage of a single image
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.inception_v3 import preprocess_input

img_path = os.path.join(test_path, 'erken', os.listdir(os.path.join(test_path, 'erken'))[0])  # any image path

img = image.load_img(img_path, target_size=(299, 299))   # same input size as in training
img_array = np.expand_dims(image.img_to_array(img), axis=0)
img_array = preprocess_input(img_array)                     # same preprocessing as in training

pred = new_model.predict(img_array)
predicted_class = CLASS_LABELS[np.argmax(pred)]
confidence = np.max(pred) * 100
print(f"Predicted stage: {predicted_class} ({confidence:.1f}% confidence)")

plt.imshow(image.load_img(img_path))
plt.title(f"Predicted: {predicted_class} ({confidence:.1f}%)")
plt.axis("off")
plt.show()
```
